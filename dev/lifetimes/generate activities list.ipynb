```python
db = bw2data.Database("ei_cutoff_3.11_remind_SSP2-NPi_2030 2025-12-15")
```

```python
import pandas as pd

rows = []

for record in db:
    row = {
        "name": record.get("name"),
        "reference product": record.get("reference product"),
    }

    for scheme, value in record.get("classifications", []):
        row[scheme] = value

    rows.append(row)

df = pd.DataFrame(rows)
```
